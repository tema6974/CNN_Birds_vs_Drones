```python
from icrawler.builtin import BingImageCrawler

queries = {
    "birds2": [
        "bird flying sky",
        "parrot flying",
        "bird wings spread"
    ],
    "drones2": [
        "drone flying sky",
        "quadcopter drone air",
        "drone aerial photography"
    ]
}

for cls, q_list in queries.items():
    for q in q_list:
        crawler = BingImageCrawler(storage={'root_dir': f'dataset/{cls}'})
        crawler.crawl(keyword=q, max_num=50)
```
